Purpose: visualize VPS13A structure and highlight missense mutation residues from CSV.

```python
import pandas as pd, py3Dmol, pathlib, re
from pathlib import Path
```

```python
variants_csv = '../data/clinical/missense-mutations.csv'
cif_path = Path("../data/structure/AF_AFQ96RL7F1.cif")
```

## Read variants

```python
variants = pd.read_csv(variants_csv)
variants
```

## Load structure

```python
if not cif_path.exists():
     raise FileNotFoundError(f"CIF file not found: {cif_path}")

cif_text = cif_path.read_text(encoding="utf-8")
```

```python
view = py3Dmol.view(width=1000, height=700)
view.addModel(cif_text, "mmcif")
#view.setStyle({"cartoon": {"color": "lightgray"}})

mut_rows = (
    variants[["aa_position", "protein_change"]]
    .dropna(subset=["aa_position"])
    .copy()
)
mut_rows = mut_rows.drop_duplicates(subset=["aa_position"])

# Base style for full structure
view.setStyle({"cartoon": {"color": "lightgray", "opacity": 0.9}})

# Highlight each mutated residue and add a label
for _, row in mut_rows.iterrows():
    pos = int(row["aa_position"])
    label = str(row["protein_change"]) if "protein_change" in mut_rows.columns else f"resi {pos}"
    sel = {"resi": str(pos)}
    view.addStyle(sel, {"stick": {"color": "red", "radius": 0.22}})
    view.addStyle(sel, {"sphere": {"color": "yellow", "radius": 0.48}})
    view.addLabel(
        label,
        {
            "position": {"resi": str(pos)},
            "backgroundColor": "white",
            "fontColor": "black",
            "fontSize": 11,
            "showBackground": True,
        },
    )

# Final camera framing
view.zoomTo()
view.show()
```
